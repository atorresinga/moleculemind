```python
from rdkit import Chem
from rdkit.Chem import Descriptors

# Our first molecule
smiles = "C(C1C(C(C(C(O1)O)O)O)O)O"

# Convert SMILES into a molecular structure
mol = Chem.MolFromSmiles(smiles)

# Inspect the molecule
print("Molecule:", smiles)
print("Number of atoms:", mol.GetNumAtoms())
print("Number of bonds:", mol.GetNumBonds())
print("Molecular weight:", Descriptors.MolWt(mol))
print("LogP:", Descriptors.MolLogP(mol))
```

```python
from rdkit import Chem
from rdkit.Chem import Descriptors

molecules = {
    "Methane": "C",
    "Ethane": "CC",
    "Ethanol": "CCO",
    "Water": "O",
    "Carbon dioxide": "O=C=O"
}

for name, smiles in molecules.items():

    mol = Chem.MolFromSmiles(smiles)

    print(f"\nMolecule: {name}")
    print("Atoms:", mol.GetNumAtoms())
    print("Bonds:", mol.GetNumBonds())
    print("Molecular Weight:", Descriptors.MolWt(mol))
    print("LogP:", Descriptors.MolLogP(mol))
```
